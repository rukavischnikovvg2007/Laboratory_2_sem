```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Создаём данные
X = pd.DataFrame({"x": [1, 2, 3, 4, 5, 6], "y_target": [0, 0, 0, 1, 1, 1]})

# Проблемный код: target внутри X И предобработка до split
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # ← здесь есть y_target!
X_train, X_test = train_test_split(X_scaled, test_size=0.3, random_state=42)

print("X колонки:", list(X.columns))
print("X_scaled shape:", X_scaled.shape)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Правильно: отделяем target от признаков
X = pd.DataFrame({"x": [1, 2, 3, 4, 5, 6]})
y = pd.Series([0, 0, 0, 1, 1, 1], name="y_target")

# 2. Split (для классификации используем stratify, чтобы классы были сбалансированы)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 3. Используем Pipeline — это защита от leakage
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(solver="liblinear"))
])

# 4. Обучаем ТОЛЬКО на train
pipe.fit(X_train, y_train)

# 5. Предсказываем на test
y_pred = pipe.predict(X_test)

print("y_test:", list(y_test))
print("y_pred:", list(y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred))
```

Правила против утечки данных (data leakage):

1. Никогда не включать target в признаки.
2. Всегда разделять данные на train/test до любой предобработки.
3. Использовать Pipeline из scikit-learn — он автоматически делает fit только на train.
4. Для временных рядов — split по времени, а не случайный.
