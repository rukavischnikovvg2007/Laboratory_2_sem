```python
import pandas as pd
import numpy as np
from pathlib import Path
from datetime import datetime
import os
import sys

print("="*60)
print("НЕДЕЛЯ 4: СОЗДАНИЕ ВИТРИНЫ ДАННЫХ (MART)")
print("="*60)
print(f"pandas: {pd.__version__}")
print("="*60)
```

```python
# Переход в корневую папку проекта
print("\n🔧 ПРОВЕРКА ПУТЕЙ")
print("-" * 40)

current = Path.cwd()
print(f"Текущая директория: {current}")

# Если мы в папке Notebooks, поднимаемся на уровень выше
if current.name == "Notebooks":
    os.chdir(current.parent)
    print(f"Перешли в корень: {Path.cwd()}")
elif (current / "data").exists():
    print("Уже в корневой папке")
else:
    # Ищем папку с проектом
    desktop = Path("C:/Users/Владимир/Desktop")
    project_path = desktop / "Laboratory_2_sem-main"
    if project_path.exists():
        os.chdir(project_path)
        print(f"Перешли в: {Path.cwd()}")
    else:
        print(f"⚠️ Не нашли проект в {project_path}")

# Проверяем
norm_dir = Path("data/normalized/variant_10")
print(f"\nПапка с normalized данными: {norm_dir}")
print(f"Существует: {norm_dir.exists()}")
```

```python
print("\n" + "="*60)
print("ШАГ 1: ЗАГРУЗКА NORMALIZED ДАННЫХ")
print("="*60)

# Проверяем папку
norm_dir = Path("data/normalized/variant_10")

if not norm_dir.exists():
    print(f"❌ ОШИБКА: Папка {norm_dir} не существует!")
    print("Сначала выполните задание 3 недели")
    sys.exit(1)

# Ищем CSV файлы
norm_files = list(norm_dir.glob("*.csv"))

if not norm_files:
    print(f"❌ ОШИБКА: В папке {norm_dir} нет CSV файлов!")
    print("Сначала выполните задание 3 недели")
    sys.exit(1)

print(f"Найдено CSV файлов: {len(norm_files)}")
for f in norm_files:
    print(f"  - {f.name}")

# Берем самый свежий
latest_norm = max(norm_files, key=lambda p: p.stat().st_mtime)
print(f"\n✅ Использую файл: {latest_norm.name}")

# Загружаем данные
df_norm = pd.read_csv(latest_norm)

print(f"\n📊 ИНФОРМАЦИЯ О ДАННЫХ:")
print(f"  - Количество строк: {df_norm.shape[0]}")
print(f"  - Количество колонок: {df_norm.shape[1]}")
print(f"  - Колонки: {list(df_norm.columns)}")

print(f"\n📋 ПЕРВЫЕ 5 СТРОК:")
print(df_norm.head())

print(f"\n📋 ПОСЛЕДНИЕ 5 СТРОК:")
print(df_norm.tail())
```

```python
print("\n" + "="*60)
print("ШАГ 2: АНАЛИЗ ДАННЫХ")
print("="*60)

# Какие страны?
print("\n🌍 СТРАНЫ В ДАННЫХ:")
countries = df_norm['country_name'].unique()
print(f"  Количество стран: {len(countries)}")
for country in countries:
    print(f"  - {country}")

print(f"\n  Коды стран (iso3): {df_norm['country_iso3'].unique()}")

# Какой индикатор?
print(f"\n📈 ИНДИКАТОР:")
print(f"  Код: {df_norm['indicator_code'].unique()[0]}")
print(f"  Название: {df_norm['indicator_name'].unique()[0]}")

# Диапазон лет
print(f"\n📅 ДИАПАЗОН ЛЕТ:")
print(f"  С {int(df_norm['year'].min())} по {int(df_norm['year'].max())}")
print(f"  Всего лет: {df_norm['year'].nunique()}")

# Проверка пропусков
print(f"\n⚠️ ПРОПУСКИ В ДАННЫХ:")
nulls = df_norm.isnull().sum()
for col in df_norm.columns:
    if nulls[col] > 0:
        print(f"  - {col}: {nulls[col]} пропусков ({nulls[col]/len(df_norm)*100:.1f}%)")
```

```python
print("\n" + "="*60)
print("ШАГ 3: СОЗДАНИЕ СПРАВОЧНИКА СТРАН")
print("="*60)

# Папка для справочников
ref_dir = Path("reference")
ref_dir.mkdir(parents=True, exist_ok=True)

# Путь к файлу
ref_file = ref_dir / "countries.csv"

# Проверяем, существует ли справочник
if ref_file.exists():
    print(f"✅ Найден существующий справочник: {ref_file}")
    df_ref = pd.read_csv(ref_file)
    print(f"   Загружено {len(df_ref)} строк")
else:
    print(f"📝 Создаем новый справочник стран...")
    
    # Создаем справочник для Германии
    countries_data = {
        "country_iso3": ["DE", "DEU"],
        "country_name_full": ["Germany", "Federal Republic of Germany"],
        "region": ["Western Europe", "Western Europe"],
        "income_group": ["High income", "High income"],
        "currency": ["Euro", "Euro"],
        "capital": ["Berlin", "Berlin"]
    }
    
    df_ref = pd.DataFrame(countries_data)
    df_ref.to_csv(ref_file, index=False)
    print(f"✅ Справочник создан: {ref_file}")

print("\n📋 СПРАВОЧНИК СТРАН:")
print(df_ref)
```

```python
print("\n" + "="*60)
print("ШАГ 4: ПРОВЕРКА КЛЮЧА ДЛЯ JOIN")
print("="*60)

join_key = "country_iso3"

print(f"🔑 Ключ соединения: '{join_key}'")

print("\n📊 КЛЮЧ В NORMALIZED ДАННЫХ:")
norm_keys = df_norm[join_key].unique()
print(f"  Уникальные значения: {norm_keys}")

print("\n📊 КЛЮЧ В СПРАВОЧНИКЕ:")
ref_keys = df_ref[join_key].unique()
print(f"  Уникальные значения: {ref_keys}")

duplicates = df_ref[join_key].duplicated().sum()
print(f"\nДубликатов в справочнике: {duplicates}")

if duplicates > 0:
    print("⚠️ Удаляем дубликаты...")
    df_ref = df_ref.drop_duplicates(subset=[join_key])
    print(f"  После удаления: {len(df_ref)} строк")
```

```python
print("\n" + "="*60)
print("ШАГ 5: ВЫПОЛНЕНИЕ JOIN")
print("="*60)

before_count = len(df_norm)
print(f"ДО join: {before_count} строк")

# Выполняем left join
df_mart = df_norm.merge(df_ref, on=join_key, how="left")

after_count = len(df_mart)
print(f"ПОСЛЕ join: {after_count} строк")

if after_count != before_count:
    print(f"⚠️ ВНИМАНИЕ! Строк стало больше на {after_count - before_count}")
else:
    print("✅ Количество строк не изменилось - отлично!")

print(f"\n📋 РЕЗУЛЬТАТ (первые 3 строки):")
print(df_mart.head(3))

# Новые колонки
new_cols = set(df_mart.columns) - set(df_norm.columns)
print(f"\n📋 НОВЫЕ КОЛОНКИ (из справочника):")
for col in new_cols:
    print(f"  - {col}")
```

```python
print("\n" + "="*60)
print("ШАГ 6: ПОДГОТОВКА ВРЕМЕННОЙ ОСИ")
print("="*60)

# Добавляем дату
df_mart['date'] = pd.to_datetime(df_mart['year'], format='%Y')

# Добавляем десятилетие
df_mart['decade'] = (df_mart['year'] // 10) * 10

print(f"✅ Добавлены колонки:")
print(f"  - date: {df_mart['date'].dtype}")
print(f"  - decade: {df_mart['decade'].dtype}")

print(f"\n📋 ПРИМЕР:")
print(df_mart[['year', 'date', 'decade']].head())
```

```python
print("\n" + "="*60)
print("ШАГ 7: РАСЧЕТ KPI (КЛЮЧЕВЫХ ПОКАЗАТЕЛЕЙ)")
print("="*60)

# Удаляем пропуски для расчета
df_clean = df_mart.dropna(subset=['value'])
print(f"Используем {len(df_clean)} строк для расчета")

# KPI 1: Средний ВВП
mean_gdp = df_clean['value'].mean()
print(f"\n📊 KPI 1: Средний ВВП за все годы")
print(f"   ${mean_gdp:,.2f}")

# KPI 2: Максимальный ВВП
max_gdp = df_clean['value'].max()
max_year = df_clean.loc[df_clean['value'].idxmax(), 'year']
print(f"\n📊 KPI 2: Максимальный ВВП")
print(f"   ${max_gdp:,.2f} (в {int(max_year)} году)")

# KPI 3: Минимальный ВВП
min_gdp = df_clean['value'].min()
min_year = df_clean.loc[df_clean['value'].idxmin(), 'year']
print(f"\n📊 KPI 3: Минимальный ВВП")
print(f"   ${min_gdp:,.2f} (в {int(min_year)} году)")

# KPI 4: Темп роста
first_year = int(df_clean['year'].min())
last_year = int(df_clean['year'].max())
value_first = df_clean[df_clean['year'] == first_year]['value'].values[0]
value_last = df_clean[df_clean['year'] == last_year]['value'].values[0]
growth = ((value_last - value_first) / value_first) * 100

print(f"\n📊 KPI 4: Темп роста {first_year} → {last_year}")
print(f"   ${value_first:,.2f} → ${value_last:,.2f}")
print(f"   Рост: {growth:.1f}%")
```

```python
print("\n" + "="*60)
print("ШАГ 8: СОХРАНЕНИЕ ВИТРИНЫ ДАННЫХ")
print("="*60)

# Создаем папку
mart_dir = Path("data/mart/variant_10")
mart_dir.mkdir(parents=True, exist_ok=True)

# Импортируем datetime прямо в ячейке (на всякий случай)
from datetime import datetime

# Сохраняем с временной меткой
timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
mart_path = mart_dir / f"mart_yearly_{timestamp}.csv"

df_mart.to_csv(mart_path, index=False, encoding='utf-8')

print(f"✅ Витрина сохранена: {mart_path}")
print(f"   Размер: {mart_path.stat().st_size} байт")
print(f"   Строк: {len(df_mart)}")
print(f"   Колонок: {len(df_mart.columns)}")

print("\n" + "="*60)
print("✅ НЕДЕЛЯ 4 ВЫПОЛНЕНА!")
print("="*60)
```
